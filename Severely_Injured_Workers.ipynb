```python
df['EventDate'] = pd.to_datetime(df['EventDate'], errors='coerce') # convert to date type
```

```python
df['Month'] = df['EventDate'].dt.month
```

```python
df['Year'] = df['EventDate'].dt.year
```

```python
print(df['Month'].value_counts().sort_index())
```

```python
print(df['Year'].value_counts().sort_index())
```

```python
# تصنيف الإصابات حسب خطورتها بناءً على كلمات في NatureTitle
severity_count = {"خطيرة": 0, "متوسطة": 0, "غير محدد": 0}

serious_keywords = ["Amputation", "Fracture", "Loss of consciousness"]

for nature in df['NatureTitle']:
    if pd.isna(nature):
        severity_count["غير محدد"] += 1
    elif any(word.lower() in str(nature).lower() for word in serious_keywords):
        severity_count["خطيرة"] += 1
    else:
        severity_count["متوسطة"] += 1

print(severity_count)
```
